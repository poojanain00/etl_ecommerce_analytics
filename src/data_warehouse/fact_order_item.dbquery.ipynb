```sql
create or replace table ecommerce_analytics.data_warehouse.fact_order_items (
order_id BIGINT,
customer_id BIGINT,
product_id STRING,
promo_id INT,
order_date_key INT,
quantity INT,
price DECIMAL(10,2),
amount DECIMAL(10,2)
);


insert into ecommerce_analytics.data_warehouse.fact_order_items
SELECT
order_number AS order_id,
customer_id,
product_id,
promo_id,
date_format(to_timestamp(order_datetime), 'yyyyMMdd') AS order_date_key,
qty as quantity,
price,
qty * price as amount
FROM batch1.data.sales_orders;

SELECT * FROM ecommerce_analytics.data_warehouse.fact_order_items;
```
